```python
import pandas as pd
```

```python
df=pd.read_csv("iris.csv")
print(df.head(10))
```

```python
#shape and datatypes of the dataset
print(df.shape)
print(df.dtypes)
```

```python
print(df.describe())
```

```python
df.info()
```

```python
#select only those rows where the petal length is more than 4.5
filtered_df=df[df["petal_length"]>4.5]
print(filtered_df)
```

```python
filtered_df=df[df["species"]=="Iris-virginica"]
print(filtered_df)
```

```python
df3=df.groupby("species")["sepal_length"].mean()
df4=df.groupby("species")["petal_width"].max()
df5=df.groupby("species")["sepal_width"].std()
print(df3), print()
print(df4), print()
print(df5)
```

```python
df["petal_ratio"]=df.apply(lambda row: row["petal_length"]/row["petal_width"], axis=1)
print(df.head())
```

```python
#average petal ratio for each species
average_petal_ratio=df.groupby("species")["petal_ratio"].mean()
print(average_petal_ratio)
```

```python
df10=pd.read_csv("titanic-dataset.csv")
```

```python
#display all columns in the dataset
print(df10.columns)
```

```python
#select passengers who are female and fare >30
filtered_df=df10[(df10["sex"]=="female")& (df10["fare"]>30)]
print(filtered_df)
```

```python
# Group by Pclass and compute survival rate, average fare, and average age
grouped_df= df10.groupby("Pclass").agg(
    Avg_surv= ("Survived", "mean"),
    Avg_fare= ("Fare", "mean"),
    Avg_age= ("Age", "mean")
)
print(grouped_df)
```
